```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

sns.set_style("whitegrid")
```

```python
def create_sample_dataset(path="data/sales_data.csv"):
    """Creates a synthetic sales dataset if it doesn't already exist."""
    os.makedirs(os.path.dirname(path), exist_ok=True)

    if os.path.exists(path):
        print(f"'{path}' already exists. Skipping creation.")
        return path

    np.random.seed(42)
    n = 200
    products = ["Product A", "Product B", "Product C", "Product D", "Product E"]
    regions = ["North", "South", "East", "West Coast", "Central"]

    data = {
        "SalesID": range(101, 101 + n),
        "Product": np.random.choice(products, n),
        "Region": np.random.choice(regions, n),
        "Sales": np.random.randint(200, 1000, n),
        "Profit": np.random.randint(20, 300, n),
        "Year": np.random.choice([2021, 2022, 2023, 2024], n),
        "Date": pd.date_range("2022-01-01", periods=n, freq="3D"),
    }

    df = pd.DataFrame(data)

    # Thoda sa missing data intentionally daal rahe hain testing ke liye
    df.loc[df.sample(frac=0.05, random_state=1).index, "Sales"] = np.nan

    df.to_csv(path, index=False)
    print(f"Sample dataset created at '{path}'")
    return path


create_sample_dataset()
```

```python
class SalesDataAnalyzer:
    """Encapsulates all sales data analysis & visualization functionality."""

    def __init__(self, file_path=None):
        self.data = None
        print("SalesDataAnalyzer object created.")
        if file_path:
            self.load_data(file_path)
```

```python
def __del__(self):
    print("SalesDataAnalyzer object destroyed. Cleaning up resources...")


SalesDataAnalyzer.__del__ = __del__
```

```python
def load_data(self, file_path):
    try:
        self.data = pd.read_csv(file_path, parse_dates=["Date"], dayfirst=False)
        print("Dataset loaded successfully!")
    except FileNotFoundError:
        print(f"Error: File '{file_path}' not found.")
    except Exception as e:
        print(f"Error while loading dataset: {e}")


SalesDataAnalyzer.load_data = load_data
```

```python
def explore_data(self):
    if self.data is None:
        print("No data loaded. Please load a dataset first.")
        return

    print("1. Display the first 5 rows")
    print("2. Display the last 5 rows")
    print("3. Display column names")
    print("4. Display data types")
    print("5. Display basic info")
    choice = input("Enter your choice: ")

    if choice == "1":
        print(self.data.head())
    elif choice == "2":
        print(self.data.tail())
    elif choice == "3":
        print(self.data.columns.tolist())
    elif choice == "4":
        print(self.data.dtypes)
    elif choice == "5":
        print(self.data.info())
        print(self.data.describe())
    else:
        print("Invalid choice.")


SalesDataAnalyzer.explore_data = explore_data
```

```python
def clean_data(self):
    if self.data is None:
        print("No data loaded. Please load a dataset first.")
        return

    print("1. Display rows with missing values")
    print("2. Fill missing values with mean")
    print("3. Drop rows with missing values")
    print("4. Replace missing values with a specific value")
    choice = input("Enter your choice: ")

    missing_rows = self.data[self.data.isnull().any(axis=1)]

    if choice == "1":
        if missing_rows.empty:
            print("No missing values found in the dataset!")
        else:
            print(missing_rows)
    elif choice == "2":
        numeric_cols = self.data.select_dtypes(include=np.number).columns
        self.data[numeric_cols] = self.data[numeric_cols].fillna(self.data[numeric_cols].mean())
        print("Missing numeric values filled with column mean.")
    elif choice == "3":
        before = len(self.data)
        self.data = self.data.dropna()
        print(f"Dropped {before - len(self.data)} rows containing missing values.")
    elif choice == "4":
        value = input("Enter the value to replace missing data with: ")
        self.data = self.data.fillna(value)
        print(f"Missing values replaced with '{value}'.")
    else:
        print("Invalid choice.")


SalesDataAnalyzer.clean_data = clean_data
```

```python
def numpy_array_operations(self, column="Sales"):
    """Convert a DataFrame column to a NumPy array and demonstrate
    creation, indexing and slicing operations."""
    if self.data is None:
        print("No data loaded. Please load a dataset first.")
        return None

    arr = self.data[column].to_numpy()

    print(f"NumPy array created from column '{column}':")
    print(arr)
    print(f"Shape: {arr.shape}, dtype: {arr.dtype}")

    print("\nIndexing -> First element:", arr[0])
    print("Indexing -> Last element:", arr[-1])
    print("Slicing -> First 5 elements:", arr[:5])
    print("Slicing -> Every 2nd element (first 10):", arr[:10:2])

    return arr


SalesDataAnalyzer.numpy_array_operations = numpy_array_operations
```

```python
def mathematical_operations(self, column="Sales"):
    """Element-wise mathematical operations on a numeric column using NumPy."""
    if self.data is None:
        print("No data loaded. Please load a dataset first.")
        return

    arr = self.data[column].to_numpy(dtype=float)

    print(f"Original '{column}' array (first 10):", arr[:10])
    print("Addition (+50):", (arr + 50)[:10])
    print("Multiplication (x1.1):", (arr * 1.1)[:10])
    print("Square root:", np.sqrt(arr)[:10])
    print("Cumulative sum (first 10):", np.cumsum(arr)[:10])


SalesDataAnalyzer.mathematical_operations = mathematical_operations
```

```python
def combine_data(self, other_dataframe, how="concat", on=None):
    """Combine the current DataFrame with another using concat, merge or join."""
    if self.data is None:
        print("No data loaded. Please load a dataset first.")
        return None

    if how == "concat":
        combined = pd.concat([self.data, other_dataframe], ignore_index=True)
    elif how == "merge":
        if on is None:
            print("Please provide a column name to merge on.")
            return None
        combined = pd.merge(self.data, other_dataframe, on=on, how="outer")
    elif how == "join":
        combined = self.data.join(
            other_dataframe.set_index(on) if on else other_dataframe,
            how="outer",
            rsuffix="_other"
        )
    else:
        print("Invalid combine method. Use 'concat', 'merge' or 'join'.")
        return None

    print(f"DataFrames combined successfully using '{how}'. New shape: {combined.shape}")
    self.data = combined
    return combined


SalesDataAnalyzer.combine_data = combine_data
```

```python
def split_data(self, by="Region"):
    """Split the DataFrame into multiple DataFrames based on a criteria column."""
    if self.data is None:
        print("No data loaded. Please load a dataset first.")
        return {}

    groups = {key: group for key, group in self.data.groupby(by)}
    print(f"Data split into {len(groups)} groups based on '{by}': {list(groups.keys())}")
    return groups


SalesDataAnalyzer.split_data = split_data
```

```python
def search_sort_filter(self):
    if self.data is None:
        print("No data loaded. Please load a dataset first.")
        return

    print("1. Search for specific sales figures/products")
    print("2. Sort data based on sales, profit or another metric")
    print("3. Filter data for specific region or time period")
    choice = input("Enter your choice: ")

    if choice == "1":
        product = input("Enter product name to search: ")
        result = self.data[self.data["Product"].str.contains(product, case=False, na=False)]
        print(result)
    elif choice == "2":
        col = input("Enter column to sort by (e.g., Sales, Profit): ")
        order = input("Ascending or Descending? (asc/desc): ").lower()
        result = self.data.sort_values(by=col, ascending=(order == "asc"))
        print(result)
    elif choice == "3":
        region = input("Enter region to filter by: ")
        result = self.data[self.data["Region"] == region]
        print(result)
    else:
        print("Invalid choice.")


SalesDataAnalyzer.search_sort_filter = search_sort_filter
```

```python
def aggregate_functions(self, group_by="Region", column="Sales"):
    """Apply aggregating functions like sum, mean, count etc."""
    if self.data is None:
        print("No data loaded. Please load a dataset first.")
        return None

    result = self.data.groupby(group_by)[column].agg(["sum", "mean", "count", "max", "min"])
    print(f"Aggregated '{column}' by '{group_by}':")
    print(result)
    return result


SalesDataAnalyzer.aggregate_functions = aggregate_functions
```

```python
def statistical_analysis(self, column="Sales"):
    """Statistical computations: std, variance, percentiles, describe()."""
    if self.data is None:
        print("No data loaded. Please load a dataset first.")
        return

    series = self.data[column].dropna()

    print(f"Statistics for '{column}':")
    print("Describe:\n", series.describe())
    print("Standard Deviation:", series.std())
    print("Variance:", series.var())
    print("25th percentile:", series.quantile(0.25))
    print("50th percentile (Median):", series.quantile(0.50))
    print("75th percentile:", series.quantile(0.75))


SalesDataAnalyzer.statistical_analysis = statistical_analysis
```

```python
def create_pivot_table(self, index="Region", columns="Product", values="Sales", aggfunc="sum"):
    """Generate a pivot table for data summarization."""
    if self.data is None:
        print("No data loaded. Please load a dataset first.")
        return None

    pivot = pd.pivot_table(
        self.data, index=index, columns=columns, values=values,
        aggfunc=aggfunc, fill_value=0
    )
    print("Pivot Table:")
    print(pivot)
    return pivot


SalesDataAnalyzer.create_pivot_table = create_pivot_table
```

```python
def visualize_data(self):
    """Menu-driven data visualization using Matplotlib and Seaborn."""
    if self.data is None:
        print("No data loaded. Please load a dataset first.")
        return None

    print("1. Bar Plot")
    print("2. Line Plot")
    print("3. Scatter Plot")
    print("4. Pie Chart")
    print("5. Histogram")
    print("6. Stack Plot")
    print("7. Seaborn Heatmap")
    print("8. Seaborn Box Plot")
    print("9. Subplots (Bar + Line + Scatter + Histogram together)")
    choice = input("Enter your choice: ")

    fig = None

    if choice == "1":
        x = input("Enter x-axis column name: ")
        y = input("Enter y-axis column name: ")
        fig, ax = plt.subplots(figsize=(8, 5))
        self.data.groupby(x)[y].sum().plot(kind="bar", ax=ax, color="skyblue")
        ax.set_title(f"{y} by {x}")
        ax.legend([y])

    elif choice == "2":
        x = input("Enter x-axis column name: ")
        y = input("Enter y-axis column name: ")
        fig, ax = plt.subplots(figsize=(8, 5))
        self.data.sort_values(x).plot(x=x, y=y, kind="line", ax=ax, marker="o")
        ax.set_title(f"{y} over {x}")

    elif choice == "3":
        x = input("Enter x-axis column name: ")
        y = input("Enter y-axis column name: ")
        print("Generating scatter plot...")
        fig, ax = plt.subplots(figsize=(8, 5))
        ax.scatter(self.data[x], self.data[y], alpha=0.6, c="teal")
        ax.set_xlabel(x)
        ax.set_ylabel(y)
        ax.set_title(f"{y} vs {x}")
        print("Scatter plot displayed successfully!")

    elif choice == "4":
        col = input("Enter column to group by for Pie chart (e.g., Region): ")
        val = input("Enter numeric column to sum (e.g., Sales): ")
        fig, ax = plt.subplots(figsize=(6, 6))
        self.data.groupby(col)[val].sum().plot(kind="pie", autopct="%1.1f%%", ax=ax)
        ax.set_ylabel("")
        ax.set_title(f"{val} share by {col}")

    elif choice == "5":
        col = input("Enter numeric column for Histogram: ")
        fig, ax = plt.subplots(figsize=(8, 5))
        ax.hist(self.data[col].dropna(), bins=15, color="orange", edgecolor="black")
        ax.set_title(f"Distribution of {col}")
        ax.set_xlabel(col)
        ax.set_ylabel("Frequency")

    elif choice == "6":
        pivot = self.data.groupby(["Year", "Region"])["Sales"].sum().unstack(fill_value=0)
        fig, ax = plt.subplots(figsize=(9, 5))
        ax.stackplot(pivot.index, pivot.T.values, labels=pivot.columns)
        ax.set_title("Sales Stack Plot by Region over Years")
        ax.set_xlabel("Year")
        ax.set_ylabel("Sales")
        ax.legend(loc="upper left")

    elif choice == "7":
        fig, ax = plt.subplots(figsize=(8, 6))
        numeric_df = self.data.select_dtypes(include=np.number)
        sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm", ax=ax)
        ax.set_title("Correlation Heatmap")

    elif choice == "8":
        y = input("Enter numeric column for Box Plot (e.g., Sales): ")
        x = input("Enter categorical column for grouping (e.g., Region): ")
        fig, ax = plt.subplots(figsize=(8, 5))
        sns.boxplot(data=self.data, x=x, y=y, ax=ax)
        ax.set_title(f"{y} distribution by {x}")

    elif choice == "9":
        fig, axes = plt.subplots(2, 2, figsize=(12, 9))
        self.data.groupby("Region")["Sales"].sum().plot(kind="bar", ax=axes[0, 0], color="skyblue")
        axes[0, 0].set_title("Sales by Region")

        self.data.sort_values("Date").plot(x="Date", y="Sales", ax=axes[0, 1])
        axes[0, 1].set_title("Sales over Time")

        axes[1, 0].scatter(self.data["Sales"], self.data["Profit"], alpha=0.6)
        axes[1, 0].set_title("Sales vs Profit")

        axes[1, 1].hist(self.data["Sales"].dropna(), bins=15, color="orange")
        axes[1, 1].set_title("Sales Distribution")

        fig.suptitle("Sales Dashboard", fontsize=16)
        fig.tight_layout()

    else:
        print("Invalid choice.")
        return None

    plt.show()
    return fig


SalesDataAnalyzer.visualize_data = visualize_data
```

```python
def save_visualization(self, fig, filename="plot.png"):
    if fig is None:
        print("No figure to save. Generate a visualization first.")
        return
    fig.savefig(filename)
    print(f"Visualization saved as {filename} successfully!")


SalesDataAnalyzer.save_visualization = save_visualization
```

```python
def main():
    analyzer = SalesDataAnalyzer()
    last_fig = None

    menu = """
========== Data Analysis & Visualization Program ==========
Please select an option:
1. Load Dataset
2. Explore Data
3. Perform DataFrame Operations
4. Handle Missing Data
5. Generate Descriptive Statistics
6. Data Visualization
7. Save Visualization
8. Exit
==============================================================
"""

    while True:
        print(menu)
        choice = input("Enter your choice: ")

        if choice == "1":
            path = input("Enter the path of the dataset (CSV file): ")
            analyzer.load_data(path)

        elif choice == "2":
            analyzer.explore_data()

        elif choice == "3":
            print("1. NumPy array operations  2. Mathematical operations  "
                  "3. Combine data  4. Split data  5. Search/Sort/Filter")
            sub = input("Enter your choice: ")
            if sub == "1":
                analyzer.numpy_array_operations()
            elif sub == "2":
                analyzer.mathematical_operations()
            elif sub == "4":
                analyzer.split_data()
            elif sub == "5":
                analyzer.search_sort_filter()
            else:
                print("Invalid choice.")

        elif choice == "4":
            analyzer.clean_data()

        elif choice == "5":
            print("1. Aggregate functions  2. Statistical analysis  3. Pivot table")
            sub = input("Enter your choice: ")
            if sub == "1":
                analyzer.aggregate_functions()
            elif sub == "2":
                analyzer.statistical_analysis()
            elif sub == "3":
                analyzer.create_pivot_table()
            else:
                print("Invalid choice.")

        elif choice == "6":
            last_fig = analyzer.visualize_data()

        elif choice == "7":
            filename = input("Enter file name to save the plot (e.g., scatter_plot.png): ")
            analyzer.save_visualization(last_fig, filename)

        elif choice == "8":
            print("Exiting the program. Goodbye!")
            break

        else:
            print("Invalid choice. Please try again.")
```

```python
main()
```
